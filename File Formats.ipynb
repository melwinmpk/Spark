### PySpark supports a wide range of file formats for reading and writing distributed data

<table>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Format</th>
        <th style="text-align: center;font-size: 2.5ex">Description</th>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">CSV</th>
        <td style="text-align: center;font-size: 2.5ex">Simple, row-based format. Easy to use but not optimized for performance.</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">JSON</th>
        <td style="text-align: center;font-size: 2.5ex">Semi-structured format. Supports nested data and flexible schema.</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Parquet</th>
        <td style="text-align: center;font-size: 2.5ex">Columnar format. Highly efficient for storage and query performance.</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">ORC</th>
        <td style="text-align: center;font-size: 2.5ex">Optimized Row Columnar. Great for Hive and Hadoop ecosystems. mainly used in Hive</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Avro</th>
        <td style="text-align: center;font-size: 2.5ex">Row-based format with schema evolution support. Good for streaming.</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Text</th>
        <td style="text-align: center;font-size: 2.5ex">Plain text files. Useful for logs or simple line-based data.</td>
    </tr>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Delta</th>
        <td style="text-align: center;font-size: 2.5ex">Transactional format from Databricks. Supports ACID and time travel.</td>
    </tr>
</table>    

### Row-Based Format VS Columnar Format



<table>
    <tr>
        <th style="text-align: center;font-size: 2.5ex">Row-Based Format</th>
        <th style="text-align: center;font-size: 2.5ex">Columnar Format</th>
    </tr>
    <tr>
        <td style="text-align: center;font-size: 2.5ex"> 
        <b>🔍 How It Works:</b>
        <ul>
            <li>Stores data row by row.</li>
            <li>All fields of a record are stored together.</li>
            <li>Example formats: CSV, JSON, Avro.</li>
       </ul>
        </td>
        <td style="text-align: center;font-size: 2.5ex">
        <b>🔍 How It Works:</b>
            <ul>
                <li>Stores data column by column.</li>
                <li>All values of a single column are stored together.</li>
                <li>Example formats: Parquet, ORC, Delta.</li>
            </ul>
        </td>
    </tr>
    <tr>
        <td style="text-align: center;font-size: 2.5ex">
        <b>✅ Use Cases:</b>
            <ul>
                <li>OLTP systems (Online Transaction Processing): Ideal for frequent inserts, updates, and deletes.</li>
                <li>Record-centric access: When you need full rows (e.g., user profiles, logs).</li>
                <li>Streaming ingestion: Easier to serialize and deserialize row-wise.</li>
            </ul>
        </td>
        <td style="text-align: center;font-size: 2.5ex">
            <b>✅ Use Cases:</b>
            <ul>
                <li>OLAP systems (Online Analytical Processing): Perfect for aggregations, filtering, and column scans.</li>
                <li>Big data analytics: Spark, Hive, and Redshift thrive on columnar formats.</li>
                <li>Compression & I/O efficiency: Similar data types compress better and reduce disk reads.</li>
            </ul>
        </td>  
    </tr>    
    <tr>
        <td style="text-align: center;font-size: 2.5ex">
        <b>⚠️ Limitations:</b>
            <ul>
<li>Slower for analytical queries that only need a few columns.</li>
<li>Less efficient compression due to mixed data types in rows</li>
            </ul>    
        </td>
        <td style="text-align: center;font-size: 2.5ex">
        <b>⚠️ Limitations:</b>
            <ul>
            <li>Not ideal for frequent row-level updates.</li>
            <li>Slightly more complex to write and serialize.</li>
            </ul>    
        </td>
    </tr>    
</table>
